````python
from tritonclient.http import InferenceServerClient
from tritonhttpclient import InferInput
import numpy as np
import json
````

config input / output
```
input [
  {
    name: "text"
    data_type: TYPE_STRING
    dims: [ -1 ] 
  }
]
output [
  {
    name: "label"
    data_type: TYPE_INT32
    dims: [ -1 ]
  },
  {
    name: "scores"
    data_type: TYPE_FP32
    dims: [ -1, -1 ]
  }
]
```

````python
# https://huggingface.co/datasets/stanfordnlp/imdb/viewer/plain_text/train?row=2
texts = [
    "I rented I AM CURIOUS-YELLOW from my video store because of all the controversy that surrounded it when it was first released in 1967.",
    "If only to avoid making this type of film in the future.",
]

request_input = InferInput(
    "text",
    [len(texts), 1],
    "BYTES"
)

input_data = np.array(texts, dtype=object)[:, np.newaxis]
request_input.set_data_from_numpy(input_data)
inputs = [request_input]
````

````python
url = "localhost:8000"
model_name = "text-classification-vllm-example"

with InferenceServerClient(url=url) as client:
    results = client.infer(model_name=model_name, inputs=inputs)
````

````python
labels = results.as_numpy("label").tolist()
print(labels)
````

````python
scores = results.as_numpy("scores").tolist()
print(scores)
````
